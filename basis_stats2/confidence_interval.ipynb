# Estimation and Confidence Intervals

**Scenario:** A manufacturer wants to estimate the mean durability of print-heads (in millions of characters printed before failure).  
**Sample Size:** 15 print-heads tested until failure.  
**Goal:** Construct 99% confidence intervals using (a) sample standard deviation and (b) known population standard deviation.

## 1. Import Libraries

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

print('Libraries imported successfully.')
```

## 2. Define the Data

```python
# Durability of 15 print-heads (in millions of characters)
data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]

n          = len(data)               # Sample size
sample_mean = np.mean(data)          # Sample mean
sample_std  = np.std(data, ddof=1)   # Sample standard deviation (ddof=1 for unbiased estimate)
std_error   = sample_std / np.sqrt(n) # Standard Error of the mean

print('Data Summary')
print('------------')
print(f'Sample Size (n)            : {n}')
print(f'Sample Mean (x-bar)        : {sample_mean:.4f} million characters')
print(f'Sample Std Dev (s)         : {sample_std:.4f} million characters')
print(f'Standard Error (s/sqrt(n)) : {std_error:.4f} million characters')
```

## Part (a): 99% Confidence Interval Using Sample Standard Deviation

### Why use the t-distribution?

When the **population standard deviation is unknown**, we use the **t-distribution** instead of the normal (z) distribution for the following reasons:

- The population standard deviation (sigma) is not known, so we estimate it using the sample standard deviation (s).
- The sample size is small (n = 15 < 30), which makes the normal approximation unreliable.
- The t-distribution has heavier tails than the normal distribution, which accounts for the additional uncertainty introduced by estimating sigma from a small sample.
- As sample size increases, the t-distribution approaches the normal distribution.

**Degrees of Freedom:** df = n - 1 = 14

**Formula:**

    CI = x-bar  +/-  t(alpha/2, df) * (s / sqrt(n))

Where:
- x-bar  = sample mean
- t      = critical value from t-distribution at 99% confidence, df = 14
- s      = sample standard deviation
- n      = sample size

```python
# --- Part (a): 99% CI using t-distribution (sample std dev) ---

confidence_level = 0.99
alpha            = 1 - confidence_level       # 0.01
df               = n - 1                      # Degrees of freedom = 14

# Critical t-value for 99% CI (two-tailed: alpha/2 in each tail)
t_critical = stats.t.ppf(1 - alpha / 2, df=df)

# Margin of Error
margin_of_error_t = t_critical * std_error

# Confidence Interval
lower_t = sample_mean - margin_of_error_t
upper_t = sample_mean + margin_of_error_t

print('Part (a): 99% Confidence Interval using t-distribution')
print('-------------------------------------------------------')
print(f'Confidence Level           : {confidence_level * 100}%')
print(f'Degrees of Freedom (df)    : {df}')
print(f'Critical t-value           : {t_critical:.4f}')
print(f'Standard Error             : {std_error:.4f}')
print(f'Margin of Error            : {margin_of_error_t:.4f}')
print()
print(f'Lower Bound                : {lower_t:.4f} million characters')
print(f'Upper Bound                : {upper_t:.4f} million characters')
print()
print(f'99% CI  =  ({lower_t:.4f},  {upper_t:.4f})')
```

### Interpretation (Part a)

We are **99% confident** that the true population mean durability of the print-heads lies between the lower and upper bounds calculated above.  
This means if we repeated this sampling process 100 times, approximately 99 of those intervals would contain the true population mean.

## Part (b): 99% Confidence Interval Using Known Population Standard Deviation

### Why use the z-distribution?

When the **population standard deviation is known**, we use the **z-distribution (standard normal distribution)** because:

- We no longer need to estimate sigma from the sample, so there is no extra uncertainty.
- The z-distribution is exact when sigma is known, regardless of sample size.

**Formula:**

    CI = x-bar  +/-  z(alpha/2) * (sigma / sqrt(n))

Where:
- x-bar  = sample mean
- z      = critical value from standard normal distribution at 99% confidence
- sigma  = known population standard deviation = 0.2
- n      = sample size

```python
# --- Part (b): 99% CI using z-distribution (known population std dev) ---

population_std = 0.2   # Known population standard deviation

# Standard Error using population std dev
std_error_z = population_std / np.sqrt(n)

# Critical z-value for 99% CI (two-tailed)
z_critical = stats.norm.ppf(1 - alpha / 2)

# Margin of Error
margin_of_error_z = z_critical * std_error_z

# Confidence Interval
lower_z = sample_mean - margin_of_error_z
upper_z = sample_mean + margin_of_error_z

print('Part (b): 99% Confidence Interval using z-distribution')
print('-------------------------------------------------------')
print(f'Confidence Level           : {confidence_level * 100}%')
print(f'Known Population Std Dev   : {population_std}')
print(f'Critical z-value           : {z_critical:.4f}')
print(f'Standard Error             : {std_error_z:.4f}')
print(f'Margin of Error            : {margin_of_error_z:.4f}')
print()
print(f'Lower Bound                : {lower_z:.4f} million characters')
print(f'Upper Bound                : {upper_z:.4f} million characters')
print()
print(f'99% CI  =  ({lower_z:.4f},  {upper_z:.4f})')
```

### Interpretation (Part b)

We are **99% confident** that the true population mean durability of the print-heads lies within the bounds calculated above using the known population standard deviation of 0.2 million characters.  
Since we used the exact population standard deviation, this interval is slightly **narrower** than the one in Part (a), which used the estimated sample standard deviation.

## 3. Comparison: t-interval vs z-interval

```python
# Side-by-side comparison
print('Comparison of Both Confidence Intervals')
print('=========================================')
print(f'{'Method':<35} {'Lower Bound':<15} {'Upper Bound':<15} {'Width'}')
print('-' * 75)
print(f'{'Part (a): t-dist (sample std)':<35} {lower_t:<15.4f} {upper_t:<15.4f} {upper_t - lower_t:.4f}')
print(f'{'Part (b): z-dist (known std)':<35} {lower_z:<15.4f} {upper_z:<15.4f} {upper_z - lower_z:.4f}')
print()
print('Observation:')
print('  The z-interval (Part b) is narrower because using the known population')
print('  standard deviation removes the extra uncertainty of estimating it from')
print('  the sample. The t-distribution compensates for this uncertainty by')
print('  producing a wider interval.')
```

```python
# Visualization: Plot both confidence intervals
fig, ax = plt.subplots(figsize=(10, 4))

# Plot intervals as horizontal lines
ax.plot([lower_t, upper_t], [1, 1], 'o-', color='steelblue', linewidth=3,
        markersize=8, label=f'Part (a) t-interval: ({lower_t:.4f}, {upper_t:.4f})')

ax.plot([lower_z, upper_z], [0.6, 0.6], 'o-', color='darkorange', linewidth=3,
        markersize=8, label=f'Part (b) z-interval: ({lower_z:.4f}, {upper_z:.4f})')

# Mark the sample mean
ax.axvline(x=sample_mean, color='green', linestyle='--', linewidth=1.5,
           label=f'Sample Mean: {sample_mean:.4f}')

ax.set_yticks([])
ax.set_xlabel('Durability (Millions of Characters)', fontsize=12)
ax.set_title('99% Confidence Intervals - Comparison', fontsize=14)
ax.legend(loc='upper right', fontsize=10)
ax.set_xlim(1.0, 1.55)
plt.tight_layout()
plt.show()
```

```python
print('FINAL SUMMARY')
print('=' * 55)
print(f'Sample Size              : {n}')
print(f'Sample Mean              : {sample_mean:.4f} million characters')
print(f'Sample Std Dev           : {sample_std:.4f} million characters')
print(f'Known Population Std Dev : 0.2000 million characters')
print()
print(f'Part (a) - t-distribution (unknown sigma):')
print(f'  Critical t-value  : {t_critical:.4f}  (df = {df})')
print(f'  Margin of Error   : {margin_of_error_t:.4f}')
print(f'  99% CI            : ({lower_t:.4f},  {upper_t:.4f})')
print()
print(f'Part (b) - z-distribution (known sigma):')
print(f'  Critical z-value  : {z_critical:.4f}')
print(f'  Margin of Error   : {margin_of_error_z:.4f}')
print(f'  99% CI            : ({lower_z:.4f},  {upper_z:.4f})')
print()
print('Conclusion:')
print('  Both intervals confirm that the average print-head durability')
print('  is approximately 1.2 million characters.')
print('  The z-interval is narrower due to using the exact population')
print('  standard deviation, while the t-interval is wider to account')
print('  for the uncertainty of estimating sigma from a small sample.')
```
